```python
from ngram_model.ngn import NgramNetwork
```

```python
network = NgramNetwork('data/hungarian_cities_3000.csv')
```

```python

chunk_size = 2
for i in range(20):
    network.train(chunk_size=chunk_size, rate=0.1)
```

```python
out = ['<begin>']
while True:
    out.append(
        network.tokenizer.decode(
            network.sample(
                network.tokenizer.encode(out[-chunk_size:])
            )
        )
    )
    

    if out[-1] == '<end>':
        break
print(''.join(out))
```
