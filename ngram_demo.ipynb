```python
from ngram_model.ngn import NgramNetwork
```

```python
network = NgramNetwork('data/hungarian_cities_3000.csv')
```

```python

for i in range(500):
    chunk_size = 2
    network.train(chunk_size=chunk_size, stop=1000, rate=1)
```

```python
out = ['<begin>']
while True:
    out.append(
        network.tokenizer.decode(
            network.sample(
                network.tokenizer.encode(out[-chunk_size:])
            )
        )
    )
    
    if out[-1] == '<end>':
        break
print(''.join(out))
```
