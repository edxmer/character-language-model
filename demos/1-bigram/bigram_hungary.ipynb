```python
import sys
import os
sys.path.append(os.path.abspath('../..'))
from src.models import bigram_network
```

```python
network = bigram_network.BigramNetwork("../../src/data/normalized/hungary_cities.txt", 100)
```

```python
k = 120
for i in range(k):
    l = network.train()
    if (i<5 or k-5 <= i):
        print(f'{i:>4}', l)
```

```python
# Predict a city name
for i in range(20):
    letters:list[str] = ['<b>']
    while True:
        sampled = network.sample(network.stoi[letters[-1]])
        letters.append(network.itos[int(sampled)])
        if (letters[-1] == '<b>'):
            break

    print(''.join(letters).replace('<b>', ''))
```
