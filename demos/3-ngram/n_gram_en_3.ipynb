```python
import sys
import os
sys.path.append(os.path.abspath('../..'))
from src.models.n_gram_network import Tokenizer, NGramNetwork
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Hyperparameters

context_size = 3
embedding_vector_size = 8
w1_out = 256
batch_size = 64
iterations = 100_000



q = 10
t = 32_000


def learning_rate_function(i:float) -> float:
    return q ** ( -i/t - 1)
```

```python
# Initializations for network

data_path = '../../src/data/normalized/en_cities.txt'
tokenizer = Tokenizer(data_path)
network = NGramNetwork(context_size, tokenizer, data_path, embedding_vector_size, w1_out)
print('Number of parameters:', sum(p.nelement() for p in network.parameters))
```

```python
# Training

network.train(batch_size, iterations, learning_rate_function)
```

```python
print('Loss on test set:', network.evaluate_test_set())
```

```python
def sample_word(net:NGramNetwork):
    context = [0] * net.CONTEXT_SIZE
    i = None
    out = []
    while i != 0:
        i = net.sample(context)
        context = context[1:] + [i]
        out.append(i)
    print(tokenizer.decode(out))
```

```python
# Sampling

for _ in range(16):
    sample_word(network)
```
