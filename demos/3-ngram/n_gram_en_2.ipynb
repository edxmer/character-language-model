```python
import sys
import os
sys.path.append(os.path.abspath('../..'))
from src.models.n_gram_network import Tokenizer, NGramNetwork
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Initializations for network

data_path = '../../src/data/normalized/en_cities.txt'
tokenizer = Tokenizer(data_path)
network = NGramNetwork(4, tokenizer, data_path, embedding_vector_size=8, w1_out=256)
print('Number of parameters:', sum(p.nelement() for p in network.parameters))
```

```python
def sample_word(net:NGramNetwork):
    context = [0] * net.CONTEXT_SIZE
    i = None
    out = []
    while i != 0:
        i = net.sample(context)
        context = context[1:] + [i]
        out.append(i)
    print(tokenizer.decode(out))
```

```python
# Training
print('\nFirst iterations with learning rate 0.1')
network.train(iterations=50_000, learning_rate_function=lambda _:0.1)

print('\nSecond iterations with learning rate 0.05')
network.train(iterations=15_000, learning_rate_function=lambda _:0.5)

print('\nThird iterations with learning rate 0.01')
network.train(iterations=15_000, learning_rate_function=lambda _:0.01)

print('\nFourth iterations with learning rate 0.005')
network.train(iterations=15_000, learning_rate_function=lambda _:0.005)

print('\nFifth iterations with learning rate 0.001')
network.train(iterations=15_000, llearning_rate_function=lambda _:0.001)
```

```python
print('Loss on test set:', network.evaluate_test_set())
```

```python
# Sampling

for _ in range(16):
    sample_word(network)
```
